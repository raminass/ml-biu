## Logistic regression: simplify the likelihood

$$
\mathcal{L}(w)=\prod_{y_{i}=1} \operatorname{Pr}\left(y=1 \mid x_{i}\right) \prod_{y_{i}=0} \operatorname{Pr}\left(y=0 \mid x_{i}\right)=
$$

$$
\begin{aligned}
\text { for any } z, z^{0}=1 & =\prod_{i=1 . . n} \operatorname{Pr}\left(y=1 \mid x_{i}\right)^{y_{i}} \prod_{i=1 . . n} \operatorname{Pr}\left(y=0 \mid x_{i}\right)^{1-y_{i}} \\
p\left(y=1 \mid \mathbf{x}_{i}\right)=\sigma\left(\mathbf{w}^{T} \mathbf{x}_{i}\right) & =\prod_{i=1 . . n} \sigma\left(w^{T} x_{i}\right)^{y_{i}} \prod_{i=1 . . n}\left(1-\sigma\left(w^{T} x_{i}\right)\right)^{1-y_{i}}
\end{aligned}
$$

- We want to find the parameter $w$ maximizing the likelihood $\mathcal{L}(w)$.
- It is more convenient to do the analysis on the logarithm of the likelihood (why?)
- Applying logarithm, does not change the maximizer(why?)
$$
\log \mathcal{L}=\sum_{i} y_{i} \log \left[\sigma\left(w^{T} x_{i}\right)\right]+\sum_{i}\left(1-y_{i}\right) \log \left[1-\sigma\left(w^{T} x_{i}\right)\right]
$$
